```python
class STRAT(Strategy):
    risk_pct = 0.02  # Risk 2% of cash per trade
    
    def next(self):
        price = self.data.Close[-1]
        
        # 1. Define your Stop Loss level (Example: 2% below entry)
        stop_loss = price * 0.98 
        sl_distance = price - stop_loss
        
        if sl_distance > 0:
            # 2. Calculate ideal size based on risk
            cash_to_risk = self.equity * self.risk_pct
            ideal_size = cash_to_risk / sl_distance
            
            # 3. Calculate maximum allowed size based on leverage (margin=0.01 -> 100x)
            # We subtract a small buffer (e.g., 2% of available cash) for commissions
            max_leverage_size = (self.equity * 0.85) / (price * self.broker.margin)
            
            # 4. Use the smaller of the two sizes, rounded down to an integer
            final_size = int(min(ideal_size, max_leverage_size))
            
            
            if final_size > 0 and not self.position:
                self.buy(size=final_size, sl=stop_loss)
```
